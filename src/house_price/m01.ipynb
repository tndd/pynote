```python
import polars as pl

df = pl.read_csv(
    "data/train.csv",
    ignore_errors=True
)
df
```

```python
(
df.null_count()
    .transpose(include_header=True)
    .sort(by="column_0", descending=True)
    .filter(pl.col("column_0") > 0)
)
```

```python
# 欠損値のある部分
df.filter(pl.col("MasVnrArea").is_null())["Id", "MasVnrArea"]
```

```python
# MasVnrAreaの欠損値は０で置き換える
df = df.with_columns(
    pl.col("MasVnrArea").fill_null(0)
)
```

```python
# 欠損値が消えてるかの確認
(
df.null_count()
    .transpose(include_header=True)
    .sort(by="column_0", descending=True)
    .filter(pl.col("column_0") > 0)
)
```

```python
# dfにおけるsalePriceとその他要素の関係を確認する
# 必要なライブラリをインポート
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 数値型の列のみを選択
numeric_features = df.select([
    col for col, dtype in df.schema.items()
    if dtype in (pl.Int64, pl.Float64)
])

# 相関係数の計算と可視化
correlation_matrix = numeric_features.corr()
plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=False,
    cmap='coolwarm',
    fmt=".1f",
    center=0,
    xticklabels=correlation_matrix.columns,
    yticklabels=correlation_matrix.columns,
    mask=np.triu(np.ones_like(correlation_matrix), k=0)
)
plt.title('Correlation Matrix with SalePrice')
plt.tight_layout()
plt.show()
```

```python
# SalePriceとの相関が強い上位5つの特徴量を抽出
corr_with_price = correlation_matrix['SalePrice']
top_features = (corr_with_price
    .abs()
    .sort(descending=True)
    .head(6)  # SalePrice自身も含まれるので6つ指定
    .drop('SalePrice')  # SalePrice自身を除外
    .index()  # 特徴量名を取得
)

# 散布図の作成
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.ravel()

for idx, feature in enumerate(top_features):
    sns.scatterplot(data=numeric_features.to_pandas(), 
                   x=feature, 
                   y='SalePrice',
                   ax=axes[idx])
    axes[idx].set_title(f'{feature} vs SalePrice')

plt.tight_layout()
plt.show()
```
